##  <font color=#bc5a45><b> Appendix 2.1: Exploratory data analysis

This section explores the structure, quality and statistical properties of the dataset to inform preprocessing and visualisation choices

```python
# Data types and missing values
print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())

# Descriptive statistics
df.describe()

# Key summary statistics
summary = df[["lifeExp", "gdpPercap", "pop"]].describe(
    percentiles=[0.25, 0.5, 0.75, 0.95]
)
display(summary)

# Unique values
print("Number of countries:", df["country"].nunique())
print("Number of continents:", df["continent"].nunique())
print("Years covered:", df["year"].min(), "-", df["year"].max())

# Duplicate check
print("Number of duplicate rows:", df.duplicated().sum())

# Outlier checks for key variables
for col in ["lifeExp", "gdpPercap", "pop"]:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[df[col] > upper_bound]

    print(f"{col} - number of potential outliers:", outliers.shape[0])

# Year distribution
df["year"].value_counts().sort_index()
```

##  <font color=#bc5a45><b> Appendix 2.3: Data Pre-processing

The dataset is complete, with no missing values or duplicate observations. Data types are appropriate, with numerical variables stored as integers or floats and categorical variables as objects - so no pre-processing required.

However GDP-per-capita and population both exhibit positively skewed distributions, with a small number of countries having exceptionally high values. This skew may distort visual interpretation; therefore, a logarithmic transformation was applied to GDP per capita to improve interpretability. Population was retained in its original form to preserve interpretability when used as a visual encoding.


```python
# Data Pre-processing

# Log transform GDP-per-capita to reduce skewness
df["log_gdpPercap"] = np.log(df["gdpPercap"])

# Compare summary statistics before and after transformation
df[["gdpPercap", "log_gdpPercap"]].describe()

# Calculate mean and median for log-transformed GDP
mean_log = df["log_gdpPercap"].mean()
median_log = df["log_gdpPercap"].median()

# Create histogram of log-transformed GDP
fig = px.histogram(
    df,
    x="log_gdpPercap",
    nbins=40,
    title="Distribution of GDP per Capita (Log Transformed)",
    labels={"log_gdpPercap": "Log GDP per Capita"}
)

# Add mean line
fig.add_vline(
    x=mean_log,
    line_dash="dash",
    line_color="red"
)

# Add legend entry for mean
fig.add_scatter(
    x=[None],
    y=[None],
    mode="lines",
    line=dict(color="red", dash="dash"),
    name="Mean"
)

# Add median line
fig.add_vline(
    x=median_log,
    line_dash="dot",
    line_color="blue"
)

# Add legend entry for median
fig.add_scatter(
    x=[None],
    y=[None],
    mode="lines",
    line=dict(color="blue", dash="dot"),
    name="Median"
)

# Improve layout and axis labels
fig.update_layout(
    height=400,
    legend_title_text="Statistics"
)

fig.update_xaxes(title="Log GDP per Capita")
fig.update_yaxes(title="Count")

fig.show()
```

Prior to transformation, the mean (≈7,215) was substantially higher than the median (≈3,532), indicating a positively skewed distribution. Following transformation, the mean and median values (≈8.16 and ≈8.17 respectively) became closely aligned, suggesting a more symmetric distribution. Additionally, the standard deviation decreased significantly, indicating reduced dispersion.”

After log transformation, the distribution becomes more symmetrical, confirming that a logarithmic scale is appropriate for subsequent analysis.”

##  <font color=#bc5a45><b> Appendix 3.1: Global health inequalities



To explore global health inequalities, a choropleth map of life expectancy is presented for 2007, the most recent year in the dataset. This visualisation highlights geographical differences in health outcomes and allows lower and higher life expectancy regions to be compared spatially.

```python
# Choropleth Map: Life Expectancy

# Select one year for the map
map_year = 2007

# Filter the dataset for the selected year
map_df = df[df["year"] == map_year].copy()

# Create choropleth map
fig = px.choropleth(
    map_df,
    locations="iso_alpha",
    color="lifeExp",
    hover_name="country",
    hover_data={
        "lifeExp": ":.1f",
        "gdpPercap": ":,.0f",
        "pop": ":,",
        "continent": True,
        "iso_alpha": False
    },
    color_continuous_scale="Viridis",
    title=f"Global Life Expectancy by Country in {map_year}",
    labels={
        "lifeExp": "Life Expectancy",
        "gdpPercap": "GDP per Capita",
        "pop": "Population",
        "continent": "Continent"
    }
)

# Improve layout
fig.update_layout(
    height=650,
    margin=dict(l=0, r=0, t=50, b=0)
)

# Display map
fig.show()
```

To improve interpretability, life expectancy was grouped into low, medium and high categories. This simplifies the continuous data and makes areas with poorer health outcomes easier to identify.

```python
# Categorise life expectancy
def classify_life_exp(life_exp):
    if life_exp < 60:
        return "Low (<60)"
    elif life_exp < 70:
        return "Medium (60–69)"
    else:
        return "High (70+)"

map_df["lifeExp_category"] = map_df["lifeExp"].apply(classify_life_exp)

# Choropleth map (categorical)
fig = px.choropleth(
    map_df,
    locations="iso_alpha",
    color="lifeExp_category",
    hover_name="country",
    hover_data={
        "lifeExp": ":.1f",
        "gdpPercap": ":,.0f",
        "pop": ":,",
        "continent": True
    },
    title="Global Life Expectancy Categories",
    category_orders={
        "lifeExp_category": [
            "Low (<60)",
            "Medium (60–69)",
            "High (70+)"
        ]
    }
)

fig.update_layout(
    height=650,
    margin=dict(l=0, r=0, t=50, b=0)
)

fig.show()
```

##  <font color=#bc5a45><b> Appendix 3.2: Economic inequalities


To examine whether economic inequalities align with health inequalities, a choropleth map of GDP per capita is presented. This visualisation allows for the geographical distribution of economic wealth to be compared with patterns observed in life expectancy. By mapping GDP per capita, regions of relative economic prosperity and deprivation can be identified, providing insight into whether wealth and health disparities occur in the same locations.

```python

# Choropleth Map: GDP per Capita (Log Scale)
map_year = 2007
map_df = df[df["year"] == map_year].copy()

fig = px.choropleth(
    map_df,
    locations="iso_alpha",
    color="log_gdpPercap",  # 👈 use log version
    hover_name="country",
    hover_data={
        "lifeExp": ":.1f",
        "gdpPercap": ":,.0f",  # still show real GDP
        "pop": ":,",
        "continent": True,
        "iso_alpha": False
    },
    color_continuous_scale="Viridis",
    title=f"Global GDP per Capita (Log Scale) in {map_year}",
    labels={
        "log_gdpPercap": "Log GDP per Capita",
        "lifeExp": "Life Expectancy",
        "pop": "Population",
        "continent": "Continent"
    }
)

fig.update_layout(
    height=650,
    margin=dict(l=0, r=0, t=50, b=0)
)

fig.show()
```

The distribution of GDP-per-capita shows clear economic disparities, with higher-income countries concentrated in regions such as Europe and North America, while lower-income countries are predominantly located in parts of Africa and Asia. When compared with the life expectancy map, a similar geographical pattern is observed, suggesting a positive association between economic wealth and health outcomes. However, this relationship is not uniform across all regions, indicating that additional factors may influence health beyond economic development alone.

##  <font color=#bc5a45><b> Appendix 3.3: Relationship between wealth and health



A scatter plot was used to visualise the relationship between all the variables (GDP-per-capita, life expectancy and population size).

```python
# Scatter plot: GDP vs Life Expectancy (2007) with trendline
df_2007 = df[df["year"] == 2007].copy()

fig = px.scatter(
    df_2007,
    x="gdpPercap",
    y="lifeExp",
    size="pop",
    color="continent",
    hover_name="country",
    log_x=True,
    size_max=120,
    trendline="ols",
    title="GDP per Capita vs Life Expectancy (2007) with Trendline"
)

# Fix axis readability
fig.update_xaxes(
    type="log",
    tickvals=[1000, 5000, 10000, 50000, 100000],
    ticktext=["1k", "5k", "10k", "50k", "100k"],
    title="GDP per Capita (log scale)"
)

fig.update_yaxes(title="Life Expectancy (years)")

# Improve visibility
fig.update_traces(opacity=0.7, marker=dict(sizemin=6))

# Clean layout
fig.update_layout(
    height=650,
    margin=dict(l=0, r=0, t=50, b=0)
)

fig.show()
```

The scatter plot reveals a clear positive relationship between GDP per capita and life expectancy, indicating that countries with higher economic output generally experience better health outcomes. The use of a logarithmic scale improves interpretability by accounting for the skewed distribution of GDP. The fitted regression line confirms this association, although the relationship is non-linear, with rapid improvements in life expectancy at lower income levels and diminishing gains at higher levels of GDP. This suggests diminishing returns, where increases in wealth beyond a certain point result in smaller improvements in health outcomes. Additionally, regional variation is evident, with African countries generally clustered at lower levels of both GDP and life expectancy, while European and Oceanian countries tend to cluster at higher levels. Population size, represented by bubble size, highlights the influence of large countries, many of which demonstrate substantial improvements in life expectancy despite lower initial income levels.

##  <font color=#bc5a45><b> Appendix 3.4: Strength of relationship



The correlation matrix visulaise the strength of those relationships, here we see a moderate positive correlation (r ≈ 0.68) between GDP-per-capita and life expectancy, supporting the relationship observed in the scatter plot. This indicates that higher levels of economic development are associated with improved health outcomes. In contrast, population shows little correlation with either variable, suggesting that population size alone does not strongly influence life expectancy or GDP per capita. While the correlation is calculated using 2007 data for clarity, subsequent analysis explores how this relationship evolves over time.

```python
# Correlation matrix (2007)
corr = df_2007[["lifeExp", "gdpPercap", "pop"]].corr()

fig = px.imshow(
    corr,
    text_auto=True,
    title="Correlation Matrix (2007)",
    color_continuous_scale="Viridis"
)
fig.show()
```

##  <font color=#bc5a45><b> Appendix 3.5: Regional inequalities



The boxplot reveals significant regional differences in life expectancy. European and Oceanian countries exhibit higher median life expectancy with relatively low variability, whereas African countries show lower median values and greater spread. This suggests that while GDP is associated with life expectancy, regional factors also play an important role in determining health outcomes

The GDP boxplot shows similar regional inequality, with Europe and Oceania generally showing higher economic values, while Africa has a lower median GDP per capita. This supports the pattern observed in the scatter plot, although the variation within continents shows that wealth alone does not fully explain health outcomes.

```python
# Boxplots: LifeExp vs Log GDP by Continent (2007)
fig = make_subplots(
    rows=1,
    cols=2,
    subplot_titles=("Life Expectancy", "Log GDP per Capita")
)

# Life Expectancy
for continent in df_2007["continent"].unique():
    fig.add_trace(
        go.Box(
            y=df_2007[df_2007["continent"] == continent]["lifeExp"],
            name=continent,
            showlegend=False
        ),
        row=1,
        col=1
    )

# Log GDP-per-Capita
for continent in df_2007["continent"].unique():
    fig.add_trace(
        go.Box(
            y=df_2007[df_2007["continent"] == continent]["log_gdpPercap"],
            name=continent,
            showlegend=False
        ),
        row=1,
        col=2
    )

fig.update_yaxes(title_text="Life Expectancy (years)", row=1, col=1)
fig.update_yaxes(title_text="Log GDP per Capita", row=1, col=2)

fig.update_layout(
    title="Regional Inequality: Health vs Wealth (2007)",
    height=600,
    width=1200
)

fig.show()
```

##  <font color=#bc5a45><b> Appendix 3.6: Country-level inequalities



To further explore inequality, ranking charts are used to identify the highest and lowest performing countries in terms of life expectancy and GDP per capita.

```python
# Ranking Charts: Top & Bottom 10 (2007)
df_2007 = df[df["year"] == 2007].copy()

# Top and bottom datasets
top_life = df_2007.nlargest(10, "lifeExp")
bottom_life = df_2007.nsmallest(10, "lifeExp")

top_gdp = df_2007.nlargest(10, "gdpPercap")
bottom_gdp = df_2007.nsmallest(10, "gdpPercap")

# Create subplot layout
fig = make_subplots(
    rows=2,
    cols=2,
    subplot_titles=(
        "Top 10 Life Expectancy",
        "Top 10 GDP per Capita",
        "Bottom 10 Life Expectancy",
        "Bottom 10 GDP per Capita"
    )
)

# Top Life Expectancy
fig.add_trace(
    go.Bar(
        x=top_life.sort_values("lifeExp")["lifeExp"],
        y=top_life.sort_values("lifeExp")["country"],
        orientation="h",
        text=top_life.sort_values("lifeExp")["lifeExp"].round(1),
        textposition="auto"
    ),
    row=1, col=1
)

# Top GDP
fig.add_trace(
    go.Bar(
        x=top_gdp.sort_values("gdpPercap")["gdpPercap"],
        y=top_gdp.sort_values("gdpPercap")["country"],
        orientation="h",
        text=top_gdp.sort_values("gdpPercap")["gdpPercap"].round(0),
        textposition="auto"
    ),
    row=1, col=2
)

#  Bottom Life Expectancy
fig.add_trace(
    go.Bar(
        x=bottom_life.sort_values("lifeExp", ascending=False)["lifeExp"],
        y=bottom_life.sort_values("lifeExp", ascending=False)["country"],
        orientation="h",
        text=bottom_life.sort_values("lifeExp", ascending=False)["lifeExp"].round(1),
        textposition="auto"
    ),
    row=2, col=1
)

# Bottom GDP
fig.add_trace(
    go.Bar(
        x=bottom_gdp.sort_values("gdpPercap", ascending=False)["gdpPercap"],
        y=bottom_gdp.sort_values("gdpPercap", ascending=False)["country"],
        orientation="h",
        text=bottom_gdp.sort_values("gdpPercap", ascending=False)["gdpPercap"].round(0),
        textposition="auto"
    ),
    row=2, col=2
)

# Layout
fig.update_layout(
    height=800,
    title="Top and Bottom 10 Countries: Life Expectancy and GDP per Capita (2007)",
    showlegend=False

)


# Axis labels (separate from layout)
fig.update_xaxes(title_text="Life Expectancy (years)", row=1, col=1)
fig.update_xaxes(title_text="GDP per Capita", row=1, col=2)
fig.update_xaxes(title_text="Life Expectancy (years)", row=2, col=1)
fig.update_xaxes(title_text="GDP per Capita", row=2, col=2)

fig.show()
```

The rankings highlight substantial inequality between countries. The top life expectancy countries are mostly high-income nations, while the lowest life expectancy countries are largely concentrated in lower-income regions. However, the top GDP countries do not perfectly match the top life expectancy countries, suggesting that wealth is strongly associated with health outcomes but does not fully explain them.

##  <font color=#bc5a45><b> Appendix 3.7: Trends over time



Using consistent colour coding across both charts allows direct comparison between health and economic trends. Life expectancy increases across all continents between 1952 and 2007, while GDP-per-capita also rises overall, although at different rates. Regional disparities remain clear, with Africa consistently showing lower average life expectancy and GDP per capita compared with Europe and Oceania. This indicates that improvements in wealth and health have occurred globally but have not been evenly distributed.

```python
# Prepare data
life_trend = df.groupby(["year", "continent"])["lifeExp"].mean().reset_index()
gdp_trend = df.groupby(["year", "continent"])["gdpPercap"].mean().reset_index()

# Use Plotly default colour palette consistently
colors = px.colors.qualitative.Plotly
continents = life_trend["continent"].unique()

fig = make_subplots(
    rows=1,
    cols=2,
    subplot_titles=("Life Expectancy", "GDP per Capita (log scale)")
)

# Life expectancy trend
for i, c in enumerate(continents):
    d = life_trend[life_trend["continent"] == c]
    fig.add_trace(
        go.Scatter(
            x=d["year"],
            y=d["lifeExp"],
            mode="lines+markers",
            name=c,
            line=dict(color=colors[i]),
            showlegend=True
        ),
        row=1,
        col=1
    )

# GDP-per-capita trend
for i, c in enumerate(continents):
    d = gdp_trend[gdp_trend["continent"] == c]
    fig.add_trace(
        go.Scatter(
            x=d["year"],
            y=d["gdpPercap"],
            mode="lines+markers",
            name=c,
            line=dict(color=colors[i]),
            showlegend=False
        ),
        row=1,
        col=2
    )

# Axis labels
fig.update_yaxes(title_text="Life Expectancy (years)", row=1, col=1)
fig.update_yaxes(title_text="GDP per Capita (log scale)", type="log", row=1, col=2)

fig.update_xaxes(title_text="Year", row=1, col=1)
fig.update_xaxes(title_text="Year", row=1, col=2)

# Layout
fig.update_layout(
    height=500,
    title="Trends in Health and Wealth Over Time",
    legend_title="Continent"
)

fig.show()
```

To compliment the line charts, side-by-side heatmaps allow comparison of health and economic trends across continents. Both life expectancy and GDP per capita increase over time, although disparities between regions persist, indicating that improvements in wealth and health are uneven.

For the GDP heatmap, colour intensity is based on log-transformed GDP to improve readability, while the cell labels show the original GDP per capita values.

```python
# Side-by-side heatmaps with values
# Prepare data
life_trend = df.groupby(["year", "continent"])["lifeExp"].mean().reset_index()
gdp_trend = df.groupby(["year", "continent"])["gdpPercap"].mean().reset_index()

life_heat = life_trend.pivot(index="continent", columns="year", values="lifeExp")
gdp_heat = gdp_trend.pivot(index="continent", columns="year", values="gdpPercap")

# Log GDP
gdp_log = np.log(gdp_heat)

fig = make_subplots(
    rows=1, cols=2,
    subplot_titles=("Life Expectancy", "GDP per Capita (Log)")
)

# Life expectancy heatmap
fig.add_trace(
    go.Heatmap(
        z=life_heat.values,
        x=life_heat.columns,
        y=life_heat.index,
        colorscale="Viridis",
        colorbar=dict(title="Life Exp", x=0.45),
        text=life_heat.round(1).values,
        texttemplate="%{text}",
    ),
    row=1, col=1
)

# GDP heatmap
fig.add_trace(
    go.Heatmap(
        z=gdp_log.values,
        x=gdp_log.columns,
        y=gdp_log.index,
        colorscale="Viridis",
        colorbar=dict(title="Log GDP", x=1.0),
        text=gdp_heat.round(0).values,         # show ORIGINAL GDP
        texttemplate="%{text}",
        textfont={"size": 10}
    ),
    row=1, col=2
)

fig.update_layout(
    title="Comparison of Health and Wealth Trends Over Time",
    height=500
)

fig.show()
```
